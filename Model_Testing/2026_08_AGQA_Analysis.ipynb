```python
import requests
import pandas as pd

base_folder_link = "https://github.com/Laxeros07/Question-Answering-about-Geometries/blob/main/Model_Testing/"

gemma_4_31_link = base_folder_link+"test_results_gemma-4-31b-it.xlsx"
glm_4_7_link = base_folder_link+"test_results_glm-4.7.xlsx"
gpt_4o_mini_link = base_folder_link+"test_results_gpt-4o-mini.xlsx"
gpt_5_4_mini_link = base_folder_link+"test_results_gpt-5.4-mini.xlsx"
gpt_5_4_nano_link = base_folder_link+"test_results_gpt-5.4-nano.xlsx"
qwen_3_5_122b_a10b_link = base_folder_link+"test_results_qwen3.5-122b-a10b.xlsx"


def read_excel_from_url(url):

  # Convert the 'blob' URL to a 'raw' URL for direct file access
  raw_url = url.replace('github.com', 'raw.githubusercontent.com').replace('/blob/', '/')

  df = pd.DataFrame()

  print(f"Reading Excel from: {raw_url}")

  try:
      # Read the Excel file directly into a pandas DataFrame
      df = pd.read_excel(raw_url)
  except Exception as e:
      print(f"Error reading Excel file: {e}")

  return df
```

```python
df_gemma_4_31 = read_excel_from_url(gemma_4_31_link)
df_glm_4_7 = read_excel_from_url(glm_4_7_link)
df_gpt_4o_mini = read_excel_from_url(gpt_4o_mini_link)
df_gpt_5_4_mini = read_excel_from_url(gpt_5_4_mini_link)
df_gpt_5_4_nano = read_excel_from_url(gpt_5_4_nano_link)
df_qwen_3_5_122b_a10b = read_excel_from_url(qwen_3_5_122b_a10b_link)
```

```python
# Combine all the loaded DataFrames into a single DataFrame
all_dfs = {
    'gemma_4_31': df_gemma_4_31,
    'glm_4_7': df_glm_4_7,
    'gpt_4o_mini': df_gpt_4o_mini,
    'gpt_5_4_mini': df_gpt_5_4_mini,
    'gpt_5_4_nano': df_gpt_5_4_nano,
    'qwen_3_5_122b_a10b': df_qwen_3_5_122b_a10b
}

# Add a 'Model' column to each DataFrame before concatenating
combined_df = pd.concat([df.assign(Model=name) for name, df in all_dfs.items()], ignore_index=True)
display(combined_df.head())
```

# One Data Frame at once

```python
# calculate all descriptive stats for a dataframe
def calculate_descriptive_stats(df):
  df_stats = df[df['Language'].notna() & (df['Language'] != 'AVERAGE')].copy()
  df_stats['Score'] = pd.to_numeric(df_stats['Score'], errors='coerce')
  descriptive_stats_by_language_nv = df_stats.groupby('Language')[['Runtime (s)', 'Score', 'Status', 'BERT_P', 'BERT_R', 'BERT_F1']].describe()
  descriptive_stats_by_language_cv = df_stats.groupby('Language')['Status'].describe(include='object')
  return descriptive_stats_by_language_nv, descriptive_stats_by_language_cv


def show_stats_by_language(all_dfs, current_df_key):

  descriptive_stats_by_language_nv, descriptive_stats_by_language_cv = calculate_descriptive_stats(all_dfs[current_df_key])

  # Define the numerical columns for which you want descriptive statistics
  selected_numeric_columns = ['Runtime (s)', 'Score', 'BERT_F1']

  # Define the descriptive statistics you want to include
  selected_stats = ['count', 'mean', 'std', 'min', 'max']

  # Select only the desired statistics for the specified columns
  desired_stats = descriptive_stats_by_language_nv.loc[:, (selected_numeric_columns, selected_stats)]

  print(f"\nDescriptive Statistics per Language (Custom Columns and Selection) for {current_df_key} (Numerical Variables Only):")
  display(desired_stats)

  print(f"\nDescriptive Statistics per Language (Custom Columns and Selection) for {current_df_key} (Categorical Variables Only):")
  display(descriptive_stats_by_language_cv)
```

```python
current_df_key = 'gpt_5_4_mini' # Explicitly define the key for the current DataFrame

calculate_descriptive_stats(all_dfs[current_df_key])

show_stats_by_language(all_dfs, current_df_key)
```

```python
import pandas as pd

# perform a row-wise comparison: give a 1 if both rows are the same (PASSED, FAILED, ERROR), 0 if not
def compute_multi_lingual_consistency(all_dfs, model_name):

  col_names =  ['model', 'EN-DE', 'EN-DE Results', 'EN-FR', 'EN-FR Results', 'DE-FR', 'DE-FR Results']

  current_df = all_dfs[model_name]

  mlc_df  = pd.DataFrame(columns = col_names)

  offset_en_de = 40
  offset_en_fr = 80

  for i in range(40):

    en_row_status = current_df.loc[i, 'Status']
    de_row_status = current_df.loc[i + offset_en_de, 'Status']
    fr_row_status = current_df.loc[i + offset_en_fr, 'Status']

    mlc_df.loc[i, 'model'] = model_name
    mlc_df.loc[i, 'Query_Type'] = current_df.loc[i, 'Query_Type']
    mlc_df.loc[i, 'EN-DE'] = int(en_row_status == de_row_status)
    mlc_df.loc[i, 'EN-FR'] = int(en_row_status == fr_row_status)
    mlc_df.loc[i, 'DE-FR'] = int(de_row_status == fr_row_status)

    mlc_df.loc[i, 'EN-DE Results'] = en_row_status+"/"+de_row_status
    mlc_df.loc[i, 'EN-FR Results'] = en_row_status+"/"+fr_row_status
    mlc_df.loc[i, 'DE-FR Results'] = de_row_status+"/"+fr_row_status

  return mlc_df
```

## Stats per language and per query

```python
# calculate all descriptive stats for a dataframe
def create_table_per_language(all_dfs, model_name):

  df = all_dfs[model_name]

  df_stats = df[df['Language'].notna() & (df['Language'] != 'AVERAGE')].copy()
  df_stats['Score'] = pd.to_numeric(df_stats['Score'], errors='coerce')

  df_stats_en = df_stats[df_stats['Language'] == 'E'].copy()
  df_stats_de = df_stats[df_stats['Language'] == 'DE'].copy()
  df_stats_fr = df_stats[df_stats['Language'] == 'FR'].copy()

  df_stats_en_nv = df_stats.groupby('Query_Type')[['Runtime (s)', 'Score', 'Status', 'BERT_P', 'BERT_R', 'BERT_F1']].describe()
  df_stats_en_cv = df_stats.groupby('Query_Type')['Status'].describe(include='object')
  return df_stats_en, df_stats_de, df_stats_fr, model_name


def calculate_descriptive_stats_per_query(df, model_name):

  ## NUMERICAL VARIABLES
  df_stats_nv = df.groupby('Query_Type')[['Runtime (s)', 'Score', 'Status', 'BERT_P', 'BERT_R', 'BERT_F1']].describe()

  # Define the numerical columns for which you want descriptive statistics
  selected_numeric_columns = ['Runtime (s)', 'Score', 'BERT_F1']

  # Define the descriptive statistics you want to include
  selected_stats = ['count', 'mean', 'std', 'min', 'max']

  # Select only the desired statistics for the specified columns
  desired_stats_nv = df_stats_nv.loc[:, (selected_numeric_columns, selected_stats)]

  ## CATEGORICAL VARIABLES = STATUS
  status_counts = df.groupby(['Query_Type', 'Status']).size().unstack(fill_value=0)

  # Calculate the total count for each Query_Type
  status_counts['Total'] = status_counts.sum(axis=1)

  # Calculate the percentage of 'PASSED' for each Query_Type
  status_counts['Percentage_PASSED'] = (status_counts['PASSED'] / status_counts['Total']) * 100

  return desired_stats_nv, status_counts, model_name




df_en, df_de, df_fr, model_name = create_table_per_language(all_dfs, MODEL_NAME)

desired_stats_nv_en, status_counts_en, model_name = calculate_descriptive_stats_per_query(df_en, model_name)
desired_stats_nv_de, status_counts_de, model_name = calculate_descriptive_stats_per_query(df_de, model_name)
desired_stats_nv_fr, status_counts_fr, model_name = calculate_descriptive_stats_per_query(df_fr, model_name)

display(desired_stats_nv_en)
display(status_counts_en)
```

## Multilingual consistency

```python
import pandas as pd

# perform a row-wise comparison: give a 1 if both rows are the same (PASSED, FAILED, ERROR), 0 if not
def compute_multi_lingual_consistency(all_dfs, model_name):

  col_names =  ['model', 'EN-DE', 'EN-DE Results', 'EN-FR', 'EN-FR Results', 'DE-FR', 'DE-FR Results']

  current_df = all_dfs[model_name]

  mlc_df  = pd.DataFrame(columns = col_names)

  offset_en_de = 40
  offset_en_fr = 80

  for i in range(40):

    en_row_status = current_df.loc[i, 'Status']
    de_row_status = current_df.loc[i + offset_en_de, 'Status']
    fr_row_status = current_df.loc[i + offset_en_fr, 'Status']

    mlc_df.loc[i, 'model'] = model_name
    mlc_df.loc[i, 'Query_Type'] = current_df.loc[i, 'Query_Type']
    mlc_df.loc[i, 'EN-DE'] = int(en_row_status == de_row_status)
    mlc_df.loc[i, 'EN-FR'] = int(en_row_status == fr_row_status)
    mlc_df.loc[i, 'DE-FR'] = int(de_row_status == fr_row_status)

    mlc_df.loc[i, 'EN-DE Results'] = en_row_status+"/"+de_row_status
    mlc_df.loc[i, 'EN-FR Results'] = en_row_status+"/"+fr_row_status
    mlc_df.loc[i, 'DE-FR Results'] = de_row_status+"/"+fr_row_status

  return mlc_df


mlc_df = compute_multi_lingual_consistency(all_dfs, MODEL_NAME)
mlc_df.describe()
```

```python
consistency_percentage = mlc_df.groupby('Query_Type')[['EN-DE', 'EN-FR', 'DE-FR']].mean() * 100

# Add a count column for each Query_Type
consistency_percentage['Count'] = mlc_df.groupby('Query_Type').size()

print("Percentage of Multilingual Consistency (1s) per Query_Type:")
display(consistency_percentage)
```

```python
# Create a new DataFrame selecting the 'count' and 'mean' statistics specifically for the 'BERT_F1' metric
# The 'BERT_F1' column in the new DataFrame will represent the mean BERT_F1 score
new_stats_summary_df = pd.DataFrame({
    'count': desired_stats_nv_en[('BERT_F1', 'count')],
    'runtime (en)': desired_stats_nv_en[('Runtime (s)', 'mean')],
    'accuracy (en)': status_counts_en['Percentage_PASSED'],
    'judge_score (en)': desired_stats_nv_en[('Score', 'mean')],
    'BERT_F1 (en)': desired_stats_nv_en[('BERT_F1', 'mean')],

    'count': desired_stats_nv_de[('BERT_F1', 'count')],
    'runtime (de)': desired_stats_nv_de[('Runtime (s)', 'mean')],
    'accuracy (de)': status_counts_de['Percentage_PASSED'],
    'judge_score (de)': desired_stats_nv_de[('Score', 'mean')],
    'BERT_F1 (de)': desired_stats_nv_de[('BERT_F1', 'mean')],

    'count': desired_stats_nv_fr[('BERT_F1', 'count')],
    'runtime (fr)': desired_stats_nv_fr[('Runtime (s)', 'mean')],
    'accuracy (fr)': status_counts_fr['Percentage_PASSED'],
    'judge_score (fr)': desired_stats_nv_fr[('Score', 'mean')],
    'BERT_F1 (fr)': desired_stats_nv_fr[('BERT_F1', 'mean')],

    'MLC (EN-DE)': consistency_percentage['EN-DE'],
    'MLC (EN-FR)': consistency_percentage['EN-FR'],
    'MLC (DE-FR)': consistency_percentage['DE-FR']

})

new_stats_summary_df = new_stats_summary_df.reset_index()

print("New DataFrame with BERT_F1 Count and Mean (Query_Type as column):")
display(new_stats_summary_df)
```
